```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

Exercise 1 Tokenization

```python
raw_text = "Dr. Johnson's class starts at 9:30 a.m.! Isn't that early?"
simple_split = raw_text.split()
print("Simple split:", simple_split)
print(f"Token count: {len(simple_split)}")

tokens = word_tokenize(raw_text)
print("Word tokenize:", tokens)
```

```python
text =  "Prof. Martinez arrived. She had a meeting at 3 p.m. It lasted 2.5 hours."

sentences = sent_tokenize(text)
for i, sent in enumerate(sentences, 1):
    print(f"Sentence {i}: {sent}")
```

Exercise 2 Stopwords

```python
stop_words = set(stopwords.words('english'))
stop_words = set(stopwords.words('english'))
print(f"Number of stopwords: {len(stop_words)}")
print(f"Sample stopwords: {list(stop_words)[:10]}")

sentence = "This is a simple example demonstrating the removal of stopwords"
tokens = word_tokenize(sentence)

print("Original:", tokens)
print("Lowercased:", [word.lower() for word in tokens])
```

Exercise 3 Stemming vs Lemmatization

```python
stemmer = PorterStemmer()
words = ['running', 'runs', 'ran', 'runner', 'easily','fairly']
print("Original → Stem")
for word in words:
    print(f"{word:12} → {stemmer.stem(word)}")
```

```python
lemmatizer = WordNetLemmatizer()
words = ['running', 'runs', 'ran', 'runner', 'easily','fairly']
print("Word         Stem      Lemma")
print("-" * 35)
for word in words:
    stem = stemmer.stem(word)
    lemma = lemmatizer.lemmatize(word)
    print(f"{word:12} {stem:9} {lemma}")
```

```python
word = "running"
print(f"\nLemmatize 'running' (default noun): {lemmatizer.lemmatize(word)}")
print(f"Lemmatize 'running' (as verb): {lemmatizer.lemmatize(word, pos='v')}")
```

Exercise 4 Complete Pipeline

```python
def preprocess_text(text, remove_stopwords=True, use_lemma=True):
    """
    Complete preprocessing pipeline
    """
    # 1. Tokenize
    tokens = word_tokenize(text.lower())

    # 2. Remove punctuation (keep only alphabetic tokens)
    tokens = [word for word in tokens if word.isalpha()]

    # 3. Remove stopwords (optional)
    if remove_stopwords:
        stop_words = set(stopwords.words('english'))
        tokens = [word for word in tokens if word not in stop_words]

    # 4. Lemmatize or stem
    if use_lemma:
        lemmatizer = WordNetLemmatizer()
        tokens = [lemmatizer.lemmatize(word) for word in tokens]
    else:
        stemmer = PorterStemmer()
        tokens = [stemmer.stem(word) for word in tokens]

    return tokens

# Test it
sample_text = """
Natural Language Processing is fascinating! It involves teaching computers
to understand human languages. Many researchers are working on this problem.
"""

result = preprocess_text(sample_text)
print("Processed tokens:", result)
print(f"Vocabulary size: {len(set(result))}")
```

```python
# Side-by-side comparison
text = "The researchers are studying natural languages using computational methods"
tokens = word_tokenize(text.lower())

# Raw
print("Raw tokens:", tokens)

# Stemmed
stemmed = [PorterStemmer().stem(w) for w in tokens]
print("Stemmed:   ", stemmed)

# Lemmatized
lemmatized = [WordNetLemmatizer().lemmatize(w, pos='v') for w in tokens]
print("Lemmatized:", lemmatized)
```
